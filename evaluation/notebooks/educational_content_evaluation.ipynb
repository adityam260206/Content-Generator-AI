## 1. Study Overview
This study evaluates three distinct AI quiz-generation strategies:
1. **Full Context:** Passing the full curriculum text directly into the LLM context prompt.
2. **Retrieval:** Splitting source text into passages, indexing, and retrieving relevant top-k chunks per topic.
3. **GenerateVerify (Generate→Verify):** Two-phase agentic pipeline where candidate items are generated and strictly verified against source passages by an independent verification critic.

### Key Research Questions & Empirical Metrics
We evaluate questions across **seven dimensions**:
- **Grounding (% Strongly Grounded)**
- **Correctness (%)**
- **Cognitive Demand (Bloom's Revised Taxonomy Index)**
- **Linguistic Clarity (%)**
- **Syllabus Coverage (%)**
- **Redundancy Overlap (%)**
- **Question Yield (%)**

```python
import json
import pandas as pd
from pathlib import Path

# Load benchmark dataset
data_path = Path('../data/benchmark_dataset.json')
with open(data_path, 'r', encoding='utf-8') as f:
    benchmark_data = json.load(f)

metrics = benchmark_data['summary_metrics']
df_summary = pd.DataFrame({
    'Strategy': list(metrics.keys()),
    'Strong Grounding (%)': [m['grounding']['strongly_grounded_pct'] for m in metrics.values()],
    'Correctness (%)': [m['correctness_pct'] for m in metrics.values()],
    'Cognitive Demand (Bloom)': [m['cognitive_demand_score'] for m in metrics.values()],
    'Clarity (%)': [m['clarity_pct'] for m in metrics.values()],
    'Coverage (%)': [m['coverage_pct'] for m in metrics.values()],
    'Redundancy (%)': [m['redundancy_pct'] for m in metrics.values()],
    'Question Yield (%)': [m['question_yield_pct'] for m in metrics.values()]
})

df_summary
```

## 2. Key Findings
- **100% Strongly Grounded Questions:** `GenerateVerify` achieved 100% strong grounding on the evaluated sample, completely eliminating ungrounded/hallucinated items.
- **Highest Cognitive Demand (1.250):** `GenerateVerify` achieved a mean cognitive demand score of **1.250**, outperforming Retrieval (1.125) and Full Context (1.050) by challenging students at Apply/Analyze levels.
- **Lowest Redundancy (4.5%):** Verification phase identified and filtered out semantically redundant question pairs.

## 3. Confounding Factors Analysis
We identified 4 primary confounding factors:
1. **Context Window Degradation ('Lost in the Middle'):** Attention drops in the middle 40% of documents in Full Context.
2. **Chunk Boundary Fragmentation:** Retrieval splitting causes multi-step reasoning drops.
3. **Domain Familiarity Bias:** Models rely on pretraining memorization on canonical concepts.
4. **Sampling Temperature:** High temperatures increase distractor plausibility but degrade grounding.

## 4. Proposed Randomized Follow-up Experiment
- **Design:** 3x3 Factorial Randomized Block Design ($N = 360$ questions).
- **Evaluator Panel:** 12 independent domain-expert raters.
- **Inter-Rater Reliability:** Measured using Cohen's Quadratic Weighted Kappa (target $\kappa \ge 0.82$).
- **Measurable Criteria:** Strict Grounding Rate, Distractor Non-Plausibility, Cognitive Complexity Index, and Normalized Student Learning Gain.
